# Python 기초 문법 복습

2학년 프로그래밍 수업에서 사용한 예제를 바탕으로 Python의 기본 요소를 다시 확인합니다.

입력과 변수 → 연산자 → 리스트 → 선택 구조 → 반복 구조 → 함수·매개변수·반환값 → 파일 입출력 순서로 실행합니다.


## 1. 입력과 변수

`input()`으로 값을 입력받고 필요한 자료형으로 변환하여 변수에 저장합니다.


```python
# 입력과 변수

value = float(input("실수를 입력하세요: "))

print("입력한 값:", value)
```

## 2. 연산자

`+`, `%`, `//`를 이용하여 계산 결과를 확인합니다.


```python
# 연산자

real_exponent = -3

stored_exponent = real_exponent + 127
remainder = 13 % 2
quotient = 13 // 2

print("저장 지수:", stored_exponent)
print("나머지:", remainder)
print("몫:", quotient)
```

## 3. 리스트

여러 개의 값을 순서대로 저장하고 `append()`로 값을 추가합니다.


```python
# 리스트

bits = []

bits.append(0)
bits.append(1)
bits.append(1)

print(bits)
```

## 4. 선택 구조

조건에 따라 서로 다른 코드를 실행합니다.


```python
# 선택 구조

number = float(input("실수를 입력하세요: "))

if number < 0:
    sign = "1"
else:
    sign = "0"

print("부호 비트:", sign)
```

## 5. 반복 구조

같은 계산을 반복하여 여러 비트를 차례로 만듭니다.


```python
# 반복 구조
# 정수를 2진수 비트 리스트로 바꾸는 과정

number = int(input("0보다 큰 정수를 입력하세요: "))
bits = []
n = number

while n > 0:
    bits.append(n % 2)
    n = n // 2

bits.reverse()

print(bits)
```

## 6. 함수 · 매개변수 · 반환값

함수는 기능을 묶어 이름을 붙인 코드입니다. 매개변수로 필요한 값을 전달하고 `return`으로 처리 결과를 돌려줄 수 있습니다.


```python
# 함수 · 매개변수 · 반환값
# 텍스트 RPG에서 사용하는 공통 함수

def clamp(value, minimum, maximum):
    return max(minimum, min(value, maximum))


hp = 120
hp = clamp(hp, 0, 100)

print(hp)
```

## 7. 파일 입출력

파일에 값을 저장하고 다시 읽어 봅니다.


```python
# 파일 입출력

f = open("새파일.txt", "w")
f.write("새 파일이 만들어졌습니다.")
f.close()

f = open("새파일.txt", "r")
print(f.read())
f.close()
```

- 텍스트 RPG에서는 이 기본 문법이 하나의 큰 프로그램 안에서 어떻게 연결되는지 읽습니다.
- IEEE 754에서는 입력·연산·리스트·조건·반복을 실제 변환 알고리즘에 적용합니다.
