## 1. 데이터 입출력

정수·실수·문자열·논릿값, 변수, 자료형 변환, 입력과 출력, 여러 값 입력을 확인합니다.


### 변수 이름과 주석

변수 이름 규칙과 `#` 주석을 확인합니다.


```python
# 변수 이름과 주석
student_score = 95
message = "Python"

# 한 줄 주석
print(student_score)
print(message)
```

## 2. 데이터 연산

산술 연산, 비교 연산, 논리 연산, 비트 단위 연산을 각각 실행하고 연산자 우선순위를 비교합니다.


```python
print(7 + 2)
print(7 - 2)
print(7 * 2)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(7 ** 2)
```

### 문자열 연산

문자열의 `+`는 연결, 문자열과 정수의 `*`는 반복에 사용합니다.


```python
print("Hello" + "Python")
print("Hi" * 3)
```

```python
n = 5
print(n < 10)
print(n > 10)
print(n == 5)
print(n != 5)
print(n <= 5)
print(n >= 5)
```

```python
n = int(input("정수 입력: "))

print(n % 2 == 0)
print(n % 3 == 0)
print((n % 2 == 0) and (n % 3 == 0))
print((n % 2 == 0) or (n % 3 == 0))
print(not (n % 2 == 0))
```

```python
a = 3
b = 5

print(a & b)
print(a | b)
print(a ^ b)
print(~a)
print(a << 1)
print(a >> 1)
```

```python
print(2 + 3 * 4)
print((2 + 3) * 4)

n = 6
print(n % 2 == 0 and n % 3 == 0)
print((n % 2 == 0) and (n % 3 == 0))
```

### 연산자 우선순위 확인

수업에서는 **거듭제곱 → 단항 연산 → 곱셈/나눗셈 계열 → 덧셈/뺄셈 → 비트 시프트 → 비트 논리 연산 → 비교 → not → and → or**의 흐름을 기준으로 식을 읽습니다. 복합식은 의도한 순서가 보이도록 괄호를 사용하는 것도 함께 확인합니다.


## 3. 제어 구조

순차 구조를 기본으로, 선택 구조와 반복 구조를 실행합니다. 선택과 반복을 중첩하거나 `break`로 반복을 끝내는 흐름도 확인합니다.


```python
score = int(input("점수 입력: "))

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("D")
```

```python
for i in range(1, 6):
    print(i)
```

```python
number = int(input("0보다 큰 정수를 입력하세요: "))
bits = []
n = number

while n > 0:
    bits.append(n % 2)
    n = n // 2

bits.reverse()
print(bits)
```

```python
n = int(input("2 이상의 정수 입력: "))
i = 2

while i < n:
    if n % i == 0:
        break
    i = i + 1

if i == n:
    print("prime")
else:
    print("composite")
```

### continue와 중첩 반복

`continue`와 `break`의 차이, 반복 구조의 중첩을 확인합니다.


```python
n = int(input("정수 입력: "))

for i in range(1, n + 1):
    if i % 3 == 0:
        continue
    print(i, end=" ")
```

```python
for row in range(1, 4):
    for col in range(1, 5):
        print(row, col)
```

내장 함수·라이브러리 함수·사용자 정의 함수를 구분하고, 정의와 호출, 매개변수, 반환값, 지역 변수와 전역 변수의 범위를 확인합니다.

### 모듈과 import

`import`, `import ... as ...`, `from ... import ...` 형태를 비교합니다.


```python
import math
print(math.sqrt(25))

import math as mt
print(mt.gcd(12, 18))

from math import sqrt
print(sqrt(36))
```

```python
def hello():
    print("Hello")
    print("Python")

hello()
```

```python
def even_sum(n):
    total = 0

    for value in range(0, n + 1, 2):
        total = total + value

    return total

print(even_sum(10))
```

```python
s = 0

def add_to_sum(n):
    global s
    k = 0

    while k < n:
        k = k + 1
        s = s + k

add_to_sum(3)
print(s)
```

## 5. 파일 처리

파일 모드를 확인하고 TXT 파일과 CSV 파일을 직접 저장하고 읽습니다.


```python
f = open("새파일.txt", "w", encoding="utf-8")
f.write("새 파일이 만들어졌습니다.")
f.close()

f = open("새파일.txt", "r", encoding="utf-8")
print(f.read())
f.close()
```

```python
f = open("memo.txt", "w", encoding="utf-8")
f.write("첫 번째 줄\n")
f.close()

f = open("memo.txt", "a", encoding="utf-8")
f.write("두 번째 줄\n")
f.close()

f = open("memo.txt", "r", encoding="utf-8")
print(f.read())
f.close()
```

```python
f = open("scores.csv", "w", encoding="utf-8")
f.write("name,score\n")
f.write("A,90\n")
f.write("B,85\n")
f.close()

f = open("scores.csv", "r", encoding="utf-8")
rows = f.read().split("\n")
f.close()

for row in rows:
    if row != "":
        print(row.split(","))
```

### readline()과 seek()

한 줄씩 읽기와 파일 위치 이동을 확인합니다.


```python
f = open("sample.txt", "w", encoding="utf-8")
f.write("first\n")
f.write("second\n")
f.close()

f = open("sample.txt", "r", encoding="utf-8")
print(f.readline().strip())
f.seek(0)
print(f.read())
f.close()
```
