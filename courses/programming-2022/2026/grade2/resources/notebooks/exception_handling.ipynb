# 예외 처리 수업 실습

프로그램 실행 중 발생할 수 있는 **예외(Exception)**를 직접 확인하고,  
`try-except`를 이용해 예외가 발생해도 프로그램이 중단되지 않도록 처리해 봅니다.

> 각 셀을 위에서부터 차례대로 실행하면서 예외가 발생하는 상황과 처리 후의 실행 흐름을 비교합니다.


## 1. IndexError

리스트의 **인덱스 범위를 벗어난 위치에 접근할 때** 발생합니다.


````python
numbers = [10, 20, 30]

print(numbers[0])
print(numbers[1])
print(numbers[2])

# 아래 줄의 주석을 지우고 실행해 보세요.
# print(numbers[3])
````

## 2. ValueError

자료형을 변환할 때 **값의 형식이 맞지 않으면** 발생할 수 있습니다.


````python
print(int("10"))

# 아래 줄의 주석을 지우고 실행해 보세요.
# print(int("값"))
````

## 3. FileNotFoundError

**읽기 모드(`r`)**로 존재하지 않는 파일을 열려고 할 때 발생합니다.

파일 열기 모드의 차이도 함께 확인합니다.

- `r` : 읽기 모드 → 파일이 없으면 `FileNotFoundError`
- `w` : 쓰기 모드 → 파일이 없으면 새로 생성
- `a` : 추가 모드 → 파일이 없으면 새로 생성


````python
# 읽기 모드(r)
# 존재하지 않는 파일을 읽으려고 하면 오류가 발생합니다.

# 아래 두 줄의 주석을 지우고 실행해 보세요.
# f = open("없는파일.txt", "r")
# f.close()
````

````python
# 쓰기 모드(w)
# 파일이 없으면 새로 만들어집니다.

f = open("새파일.txt", "w")
f.write("새 파일이 만들어졌습니다.")
f.close()

print("새파일.txt 생성 완료")
````

````python
# 방금 만든 파일을 읽기 모드(r)로 열어 봅니다.

f = open("새파일.txt", "r")
print(f.read())
f.close()
````

````python
# 추가 모드(a)
# 기존 내용 뒤에 새로운 내용을 덧붙입니다.

f = open("새파일.txt", "a")
f.write("\n내용을 추가했습니다.")
f.close()

f = open("새파일.txt", "r")
print(f.read())
f.close()
````

# 5. try-except로 예외 처리하기

기본 구조는 다음과 같습니다.

```python
try:
    예외가 발생할 수 있는 코드

except 예외종류:
    예외가 발생했을 때 실행할 코드
```

`except 예외종류:`는 **지정한 종류의 예외가 발생했을 때** 실행됩니다.


````python
try:
    age = int(input("나이를 숫자로 입력하세요: "))
    print("입력한 나이:", age)

except ValueError:
    print("숫자로 입력하세요.")
````

# 6. 여러 예외를 각각 처리하기

예외의 종류에 따라 **서로 다른 처리를 하고 싶을 때** `except`를 여러 개 작성할 수 있습니다.


````python
try:
    num = int(input("나눌 숫자를 입력하세요: "))
    result = 100 / num
    print("결과:", result)

except ValueError:
    print("숫자를 입력해야 합니다.")

except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
````

# 7. 여러 예외를 한 번에 처리하기

여러 예외가 발생했을 때 **같은 처리를 하고 싶다면** 괄호 안에 예외 종류를 함께 작성할 수 있습니다.


````python
try:
    num = int(input("나눌 숫자를 입력하세요: "))
    result = 100 / num
    print("결과:", result)

except (ValueError, ZeroDivisionError):
    print("입력값을 확인하세요.")
````

## 핵심 정리

- `IndexError` : 리스트의 인덱스 범위를 벗어났을 때
- `ValueError` : 자료형 변환 시 값의 형식이 맞지 않을 때
- `FileNotFoundError` : 읽기 모드(`r`)로 존재하지 않는 파일을 열려고 할 때
- `ZeroDivisionError` : 0으로 나눌 때
- `try` : 예외가 발생할 수 있는 코드를 실행
- `except` : 지정한 예외가 발생했을 때 처리
- `except (예외1, 예외2):` : 여러 예외를 같은 방법으로 처리
